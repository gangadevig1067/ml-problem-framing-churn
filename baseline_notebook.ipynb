```python
import io
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix

# 1. Load dataset directly (self-contained so it runs without network errors)
csv_data = """customer_id,tenure_months,support_tickets,monthly_spend_inr,last_login_days,plan_type,churned
C001,2,4,499,21,Basic,1
C002,18,1,1299,2,Pro,0
C003,6,3,799,14,Basic,1
C004,30,0,1499,1,Pro,0
C005,11,2,999,5,Standard,0
C006,1,5,499,30,Basic,1
C007,24,1,999,3,Standard,0
C008,8,4,799,17,Basic,1
C009,15,2,1299,4,Pro,0
C010,4,3,499,12,Basic,1
C011,20,0,999,2,Standard,0
C012,7,2,799,9,Standard,0"""

df = pd.read_csv(io.StringIO(csv_data))

print("Dataset Shape:", df.shape)
print("\nMissing Values:\n", df.isnull().sum())
print("\nTarget Class Balance (churned):\n", df['churned'].value_counts())

# 2. Apply Non-ML Baseline Rule (Flag if last_login_days > 10)
df['baseline_pred'] = (df['last_login_days'] > 10).astype(int)

# 3. Evaluate Baseline Performance
acc = accuracy_score(df['churned'], df['baseline_pred'])
prec = precision_score(df['churned'], df['baseline_pred'])
rec = recall_score(df['churned'], df['baseline_pred'])
cm = confusion_matrix(df['churned'], df['baseline_pred'])

print(f"\n--- Non-ML Baseline Results ---")
print(f"Accuracy:  {acc:.2f}")
print(f"Precision: {prec:.2f}")
print(f"Recall:    {rec:.2f}")
print("\nConfusion Matrix:\n", cm)
print("\nCustomer-Level Comparison:")
print(df[['customer_id', 'plan_type', 'last_login_days', 'support_tickets', 'churned', 'baseline_pred']])
```
